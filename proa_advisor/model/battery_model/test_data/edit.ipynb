```python
import numpy as np
import pandas as pd
import os
```

```python
file_list = []
for file in os.listdir():
    if file.endswith(".csv"):
        file_list.append(file)
try:
    file_list.sort(key=lambda x: int(x.split("_")[-1].split(".")[0]))
except ValueError:
    print("Error: One or more filenames do not follow the expected format.")
print(file_list)
print(f"Total {len(file_list)} files found.")
```

```python
MAX_ROWS_PER_FILE = 500000
```

```python
def split_df(df, max_rows=MAX_ROWS_PER_FILE):
    df_list = []
    for i in range(0, len(df), max_rows):
        df_list.append(df.iloc[i:i + max_rows])
    return df_list
```

```python
print("Total files to process:", len(file_list))
for i, file in enumerate(file_list):
    df = pd.read_csv(file)
    if len(df) > MAX_ROWS_PER_FILE:
        df_list = split_df(df)
        print(f"File {file} has {len(df)} rows, splitting into {len(df_list)} parts.")
        for j, split_df in enumerate(df_list):
            split_df.to_csv(f"{file.split('.')[0]}_part{j+1}.csv", index=False)
            print(f"Created {file.split('.')[0]}_part{j+1}.csv with {len(split_df)} rows.")
        print(f"File {file} split into {len(df_list)} parts.")
        os.remove(file)
```

```python
for file in file_list:
    df = pd.read_csv(file)
    if df['ch6'][0] < 40000:
        df['ch6'] *= 2
    df['ch6'] -= 1000
        
    df.to_csv(file, index=False)
    #print(df.head())
    print(f"{file} has been edited.")
    
```
